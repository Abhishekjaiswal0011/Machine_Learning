```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


import warnings
warnings.filterwarnings("ignore")
```

```python
df = sns.load_dataset('mpg')
```

```python
df
```

```python
df.drop("name",axis = 1,inplace = True)
```

```python
df.head()
```

```python
df.info()
```

```python
df.isna().sum()
```

```python
df.dtypes
```

```python
df.shape
```

```python
# since we have not done outlier treatment , then better iddea would be replace the missing value with median
```

```python
df['horsepower'].median()
```

```python
df['horsepower'] = df['horsepower'].fillna(df['horsepower'].median())
```

```python
df.isna().sum()
```

```python
df.info()
```

```python
df.dtypes
```

```python
df['origin'].value_counts()
```

```python
# data encoding
df["origin"] = df["origin"].map({"usa":1, "japan":2, "europe":3})
```

```python
df["origin"]
```

```python
df.dtypes
```

```python
df["origin"] = df["origin"].astype(int)
```

```python
df.dtypes
```

```python
# separate into X and y
```

```python
X = df.drop('mpg',axis = 1)
y = df['mpg']
```

```python
X
```

```python
y
```

```python
# train test splt

from sklearn.model_selection import train_test_split
X_train,X_test,y_train, y_test = train_test_split(X,y,test_size = 0.3,random_state = 1)
```

```python
X_train
```

```python
X_test
```

```python
X_train.shape
```

```python
#Simple Linear REgression
from sklearn.linear_model import LinearRegression
regression_model = LinearRegression()
```

```python
regression_model
```

```python
regression_model.fit(X_train,y_train)
```

```python
regression_model.coef_
```

```python
for i , col_name in enumerate(X_train.columns):
    print(f"The Coefficient for{col_name} is {regression_model.coef_[i]}")
```

```python
# observations:
# coefficints are relativelt smaller, if one independent variable changes
#there willl be not much difference in prediction
# this is sometione called as smother model

# these features might not be contributing in model training
```

```python
from sklearn.metrics import r2_score

y_pred_linear = regression_model.predict(X_test)
r2_linear = r2_score(y_test,y_pred_linear)
print(f"The square of linear regression {r2_linear}")
```

```python
# Regularised model
# ridge regression
```

```python
from sklearn.linear_model import Ridge
ridge_regression_model = Ridge(alpha = 0.1)
ridge_regression_model


# in practical impimentaion lambda is alpha
```

```python
ridge_regression_model.fit(X_train,y_train)
```

```python
for i , col_name in enumerate(X_train.columns):
    print(f"The Coefficient for{col_name} is {ridge_regression_model.coef_[i]}")
```

```python
# for ridge regression evaluation
y_pred_ridge = ridge_regression_model.predict(X_test)
r2_ridge = r2_score(y_test,y_pred_ridge)
print(f"The r_square of ridge regression {r2_linear}")
```

```python
# we dont see much variation in coeff of ridge regression as compared to linear Regression
```

```python
from sklearn.linear_model import Lasso
lasso_regression_model = Lasso(alpha = 0.5)
lasso_regression_model
```

```python
lasso_regression_model.fit(X_train,y_train)
for i , col_name in enumerate(X_train.columns):
    print(f"The Coefficient for{col_name} is {lasso_regression_model.coef_[i]}")
```

```python
# three features coefficient is 0, lasso helps in features selection
```

```python
y_pred_lasso = lasso_regression_model.predict(X_test)
r2_lasso = r2_score(y_test,y_pred_lasso)
print(f"The r_square of lasso regression {r2_lasso}")
```

```python
# Elastic net
from sklearn.linear_model import ElasticNet
elastic_net_model =  ElasticNet(alpha = 1,l1_ratio = 0.5)
elastic_net_model.fit(X_train,y_train)
```

```python
for i , col_name in enumerate(X_train.columns):
    print(f"The Coefficient for{col_name} is {elastic_net_model.coef_[i]}")
```

```python
y_pred_elastic = elastic_net_model.predict(X_test)
r2_elastic = r2_score(y_test, y_pred_elastic)
print(f"The R2 score of Elastic Net regression: {r2_elastic}")
```

```python
#CV with hyperparameter tuning
```

```python
from sklearn.model_selection import GridSearchCV,RandomizedSearchCV
from sklearn.linear_model import Lasso,Ridge
```

```python
# gridsearhcv for Lasso
#difine the model
lasso = Lasso()
```

```python
lasso
```

```python
# grid searchcv>> 


param_grid = {'alpha': [0.001,0.01,0.1,1,10,100]}
param_grid
```

```python
grid_search = GridSearchCV(estimator = lasso,param_grid = param_grid,cv = 5,scoring = 'r2',verbose = 2)
```

```python
grid_search
```

```python
grid_search.fit(X_train,y_train)
```

```python
grid_search.best_params_
```

```python
grid_search.best_params_['alpha']
```

```python
grid_search.best_score_ # mean cross- validated score of the best_estimator
```

```python

grid_search.best_estimator_
```

```python

y_pred = grid_search.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
# randomized search cv for Lasso
```

```python
lasso = Lasso()
param_distributions = {'alpha':[0.001,0.01,0.1,10,1,100]}
random_search_lasso = RandomizedSearchCV(estimator = lasso, param_distributions = param_distributions, n_iter=3,  cv = 5,scoring = 'r2',verbose = 2)
random_search_lasso
```

```python
random_search_lasso.fit(X_train,y_train)
# due to n_iter parameters in each cross validation random 3 parameter selected
```

```python
random_search_lasso.best_estimator_
```

```python
random_search_lasso.best_score_
```

```python
y_pred = random_search_lasso.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Ridge grid Search CV
ridge = Ridge()
param_grid = {'alpha': [0.001,0.01,0.1,1,10,100]}
ridge = GridSearchCV(estimator = ridge,param_grid = param_grid,cv=5,scoring='r2',verbose = 2)
ridge.fit(X_train,y_train)
print(ridge.best_params_)
print(ridge.best_estimator_)
y_pred = ridge.best_estimator_.predict(X_test)
print(r2_score(y_test,y_pred))

```

```python
ridge.best_estimator_
```

```python
y_pred = ridge.best_estimator_.predict(X_test)
r2_score(y_test,y_pred)
```

```python
#Ridge  grid Search CV
ridge = Ridge()
param_distributions = {'alpha': [0.001,0.01,0.1,1,10,100]}
ridge = RandomizedSearchCV(estimator = ridge,param_distributions = param_grid,n_iter = 2,cv=5,scoring='r2',verbose = 2)
ridge.fit(X_train,y_train)
print(ridge.best_params_)
print(ridge.best_estimator_)
y_pred = ridge.best_estimator_.predict(X_test)
print(r2_score(y_test,y_pred))

```

```python
# Elastic net  grid Search CV
model = ElasticNet()
param_distributions = {'alpha': [0.001,0.01,0.1,1,10,100],
                      'l1_ration': [0.1,0.4,0.9]}
model = GridSearchCV(estimator = model,param_grid = param_grid,cv=5,scoring='r2',verbose = 2)
model.fit(X_train,y_train)
print(model.best_params_)
print(model.best_estimator_)
y_pred = model.best_estimator_.predict(X_test)
print(r2_score(y_test,y_pred))

```

```python
# Elastic net Randomized Search CV
model = ElasticNet()
param_distributions = {'alpha': [0.001,0.01,0.1,1,10,100],
                      'l1_ratio': [0.1,0.4,0.9]}
model = RandomizedSearchCV(estimator = model,param_distributions=param_distributions,cv=5,scoring='r2',verbose = 2)
model.fit(X_train,y_train)
print(model.best_params_)
print(model.best_estimator_)
y_pred = model.best_estimator_.predict(X_test)
print(r2_score(y_test,y_pred))

```
