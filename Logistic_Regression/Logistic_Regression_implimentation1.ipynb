```python
# Logistic regression >>till now we have studied with twoo values of the target variable>>binary classification
```

```python
df = pd.DataFrame(data.data,columns = data.feature_names)
df['target'] = data.target
```

```python
df
```

```python
df.head()
```

```python
df.tail()
```

```python
df.sample(1)
```

```python
df.target.unique()
```

```python
# for binary classification >>only two class needed so we will filter out class 2
df = df[df['target'] != 2]
```

```python
df.target.unique()
```

```python
df
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
X
```

```python
y
```

```python
# train test split
from sklearn .model_selection import train_test_split
X_train,X_test , y_train,y_test = train_test_split(X,y, test_size = 0.20,random_state = 1)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
X_train
```

```python
# model training
from sklearn .linear_model import LogisticRegression
```

```python
classifier = LogisticRegression()
classifier
```

```python
classifier.fit(X_train,y_train)
```

```python
y_pred = classifier.predict(X_test)
y_pred
```

```python
classifier.predict_proba(X_test) # corresponding to higher probability class is predicted ,cutoff is 0.5
```

```python
# Evaluation metrics
from sklearn.metrics import confusion_matrix,accuracy_score,classification_report
```

```python
confusion_matrix(y_test,y_pred)
```

```python
accuracy_score(y_test,y_pred)
```

```python
print(classification_report(y_test,y_pred))
```

```python
# first two rows , 0 and 1 >> presion recall and,f1 score and no of data points with respective class
# macro avg  >> for each class take simple avg>> avg precision of class o and class 1 will be macro avg 
# similarly  for recall f1 score
# support will be total no of datapoints

# weighted avg is used in case of class imbalance ,it gives importance to performance of dominant class 
```

```python
from sklearn.metrics import roc_curve, auc
# get predicted probabilities of class 1
classifier .predict_proba(X_test) [:, 1]
```

```python
y_pred_proba = classifier .predict_proba(X_test) [:, 1]
```

```python
y_pred_proba
```

```python
# receiver operating characterstics curve
# inputs : y_test , y_pred_proba. it will return TPR,FDR with different cutoff of probabilities
fpr,tpr,thresholds = roc_curve(y_test , y_pred_proba)
```

```python
thresholds
```

```python
tpr
```

```python
fpr
```

```python
roc_auc = auc(fpr,tpr)
```

```python
#plot roc-auc curve

plt.figure(figsize=(8, 6))  # Create a new figure with a specified size
plt.plot(fpr, tpr, color='darkorange', linewidth=2, label='ROC curve (area = %0.2f)' % roc_auc)  # Plot ROC curve,
plt.plot([0, 1], [0, 1], color='navy', linewidth=2, linestyle='--')  # Plot the diagonal line representing random guessing
plt.xlim([0.0, 1.0])  # Set x-axis limits
plt.ylim([0.0, 1.05])  # Set y-axis limits
plt.xlabel('False Positive Rate')  # Set x-axis label
plt.ylabel('True Positive Rate')  # Set y-axis label
plt.title('Receiver Operating Characteristic (ROC) Curve')  # Set plot title
plt.legend(loc="lower right")  # Add legend to the plot
plt.show()  # Show the plot
```
