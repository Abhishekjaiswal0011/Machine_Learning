```python
# MLR>> model the relationship betwween dependent variable y and more than one independent variable
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
from sklearn.datasets import load_diabetes
```

```python
diabetes = load_diabetes()
```

```python
diabetes
```

```python
print(diabetes.DESCR)
```

```python
# to predict desease progression one year after the baseline
```

```python
diabetes.data
```

```python
diabetes.target
```

```python
diabetes.feature_names
```

```python
data = pd.DataFrame(diabetes.data, columns = diabetes.feature_names)
```

```python
data
```

```python
data['target'] = diabetes.target
```

```python
data['target'] 
```

```python
#EDA and data preparation

X = data.drop('target',axis = 1)
y = data['target']
```

```python
X
```

```python
y
```

```python
#  split the data into train and test
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y ,test_size = 0.2,random_state = 1)
```

```python
X_train
```

```python
X_test
```

```python
y_train
```

```python
y_test
```

```python
X_train.shape,X_test.shape
```

```python
# scaling optional so we dont do here
```

```python
#model training 
from sklearn.linear_model import LinearRegression
model = LinearRegression()
```

```python
model.fit(X_train,y_train)
```

```python
X_train.shape
```

```python
model.coef_
```

```python
len(model.coef_)
```

```python
model.intercept_
```

```python
# interpretaion
# say age has coef value -30.62219046
# interpretaion>> 1 unit change in age than target value decrease by -30 units on an average
# when all  other features/factors are kept constant
X_train
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```
