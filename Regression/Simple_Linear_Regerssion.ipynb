```python
diabetes = load_diabetes()
```

```python
data = pd.DataFrame(diabetes.data,columns = diabetes.feature_names)
```

```python
data
```

```python
data.head()
```

```python
data['target'] = diabetes.target
```

```python
data
```

```python
#EDA, DATACLEANING,DATAPREPARATION,FEATURE ENGINEERING BUT WE HAVE DONE ALL THAT BEFORE
```

```python
X = data[["bmi"]] #independent variable 
Y = data['target'] # dependent variable
```

```python
X
```

```python
Y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, Y_train, Y_test = train_test_split(X,Y, test_size = 0.2,random_state = 2)
# random_state when fixed helps us to reproduce the same result again and again
```

```python
X_train.shape
```

```python
X_test
```

```python
Y_train
```

```python
Y_test
```

```python
 # model training
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model
```

```python
X_train
```

```python
Y_train
```

```python
# training of model
model.fit(X_train,Y_train)
# at the backend it will use gradient descent to give optional coeffficient
```

```python
model.coef_ #attribute of the model
```

```python
model.intercept_
```

```python
y_pred = model.predict(X_test)
```

```python
y_pred
```

```python
# visualize the result
plt.scatter(X_test,Y_test,color = 'black',label = "Actual")
plt.plot(X_test,y_pred,color = 'blue',linewidth = 3,label = "Predicted linear regression line")
plt.xlabel("BMI")
plt.ylabel("One year pregression target")
plt.title("Linear regression on diabetes data")
plt.legend()
plt.show()
```

```python
# model evaluation
from sklearn.metrics import mean_squared_error
mean_squared_error(Y_test,y_pred)
```
